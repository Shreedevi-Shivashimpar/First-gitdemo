4. find factorial of a number

```python
n=int(input("Enter a number:"))
def fact(n):
  if n==0:
    return 1
  return fact(n-1)*n
print(fact(n))
```

5. Fibonacci series

```python

def fib(n):
  if n<=1:
    return n
  else:
    return fib(n-1)+fib(n-2)
n=int(input("Enter a terms:"))
if n==0:
  print("Enter a valid number")
else:
  for i in range(n):
    print(fib(i),end=" ")
```

6. calculate x^n(power of a number)

```python
def power(x,n):
  if n==0:
    return 1
  return x*power(x,n-1)
x=int(input("Enter a number:"))
n=int(input("Enter a power:"))
print(power(x,n))
```

7. count the number of digits of a number

```python
n=int(input("Enter a number:"))
def count(n):
  if n==0:
    return 0
  return 1+count(n//10)
print(count(n))
```

8. sum of the digits of a number

```python
n=int(input("Enter a number:"))
def sum(n):
  if n==0:
    return 0
  return n%10 + sum(n//10)
print(sum(n))
```

9. to reverse a given number



```python
n=int(input("Enter a number"))
def reverse(n):
  if n==0:
    return 0
  else:
    l=len(str(n))
    digit=n%10
    rno=n//10
    return digit*(10**(l-1))+reverse(rno)
print(reverse(n))
```
